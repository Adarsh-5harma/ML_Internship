Plotting a Vector

```python
# quiver is a function for arrows 
plt.quiver(0,0,4,5)
plt.show()
```

```python
plt.quiver(0,0,4,5,scale_units = 'xy', angles='xy', scale = 1, color = 'r')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

```python
plt.quiver(6,4,scale_units = 'xy', angles='xy', scale = 1, color = 'r')
plt.quiver(-4,-5,scale_units = 'xy', angles='xy', scale = 1, color = 'b')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

Addition of two Vectors

```python
vector_1 = np.asarray([2,3])

vector_2 = np.asarray([3,-2])

sum_of_vector = vector_1 + vector_2
sum_of_vector
```

```python
plt.quiver(2,3,scale_units = 'xy', angles='xy', scale = 1, color = 'r')
plt.quiver(3,-2,scale_units = 'xy', angles='xy', scale = 1, color = 'b')
plt.quiver(5,1,scale_units = 'xy', angles='xy', scale = 1, color = 'y')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

```python
difference = vector_1 - vector_2
difference
```

```python
plt.quiver(2,3,scale_units = 'xy', angles='xy', scale = 1, color = 'r')
plt.quiver(3,-2,scale_units = 'xy', angles='xy', scale = 1, color = 'b')
plt.quiver(-1,5,scale_units = 'xy', angles='xy', scale = 1, color = 'y')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

Multiplying Vector by a Scalar

```python
vector_1 = np.asarray([2,3])

multiplied_vector = 3 * vector_1

multiplied_vector
```
