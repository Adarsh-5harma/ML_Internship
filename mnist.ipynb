```python
import matplotlib.pyplot as plt

def plot_digit(image_data):
    image = image_data.reshape(28,28)
    plt.lmshow(image, cmap="binary")
    plt.axis("off")
    some_digit= X[0]
    plot_digit(some_digit)
    plt.show()
```
